```python
import os
import hnswlib

# Create empthy index
index = hnswlib.Index(space='cosine', dim=384)

# Define hnswlib index path
index_path = './qa_hnswlib.index'

# Load index if available
if os.path.exists(index_path):
    print('Loading index...')
    index.load_index(index_path)
# Else index data collection
else:
    # Initialise the index
    print('Started creating HNSWLIB index')
    index.init_index(max_elements=corpus_embeddings.size(0), ef_construction=400, M=64)
    #  Compute the HNSWLIB index (it may take a while)
    index.add_items(corpus_embeddings.cpu(), list(range(len(corpus_embeddings))))
    # Save the index to a file for future loading
    print(f'Saving index to: {index_path}')
    index.save_index(index_path)
```

```python
# %%

!pip install -q sentence-transformers datasets==2.16.0
from sentence_transformers import SentenceTransformer, util
import datasets
from datasets import load_dataset
import torch
EMBEDDING_MODEL = 'all-MiniLM-L6-v2'  # Compact and fast model
SIMILARITY_THRESHOLD = 0.6
TOP_K = 3  # Number of most similar questions to retrieve

model = SentenceTransformer(EMBEDDING_MODEL)

dataset = datasets.load_dataset("FreedomIntelligence/RAG-Instruct", split="train")

# Extract questions from the dataset
questions_dataset = [item["question"] for item in dataset]
print(f"✅ {len(questions_dataset)} questions loaded.")

# Generate embeddings for all dataset questions
embeddings_dataset = model.encode(questions_dataset, convert_to_tensor=True)

# ========================
# 🔍 RETRIEVAL + PROMPT FUNCTION
# ========================
def get_context_and_prompt(user_question):
    # Encode user input
    user_embedding = model.encode(user_question, convert_to_tensor=True)

    # Compute cosine similarity with dataset questions
    similarities = util.pytorch_cos_sim(user_embedding, embeddings_dataset)[0]

    # Get top K most similar question indices
    top_k_indices = torch.topk(similarities, k=TOP_K).indices.tolist()
    top_k_scores = similarities[top_k_indices].tolist()

    # If the most similar question is not relevant enough
    if top_k_scores[0] < SIMILARITY_THRESHOLD:
        return {
            "llm_response": "I do not have enough information to complete this task.",
            "similarities": top_k_scores,
            "similar_questions": [questions_dataset[i] for i in top_k_indices]
        }

    # Collect context documents from top-k similar questions
    context_documents = []
    for idx in top_k_indices:
        contexts = dataset[idx]["context"]
        if isinstance(contexts, list):
            context_documents.extend(contexts)
        else:
            context_documents.append(contexts)

    # Build prompt for the LLM
    prompt = f"""Answer the following question using only the information provided in the context.
If you cannot find enough information, respond: "I do not have enough information to complete this task."

Question: {user_question}

Context:
{chr(10).join(context_documents)}
"""

    return {
        "prompt": prompt,
        "similarities": top_k_scores,
        "similar_questions": [questions_dataset[i] for i in top_k_indices]
    }

# ========================
# 🧪 INTERACTIVE TEST
# ========================
user_question = input("Enter your question: ")

result = get_context_and_prompt(user_question)

if "llm_response" in result:
    print("\n⚠️ Suggested response (insufficient context):")
    print(result["llm_response"])
else:
    print("\n📝 Prompt generated for the LLM:\n")
    print(result["prompt"])
    print("\n🔎 Similarities:", result["similarities"])
    print("📌 Similar questions:", result["similar_questions"])
```

```python
# %%
!pip install -q sentence-transformers datasets==2.16.0

from sentence_transformers import SentenceTransformer, CrossEncoder, util
from datasets import load_dataset
import torch
EMBEDDING_MODEL = 'all-MiniLM-L6-v2'
CROSSENCODER_MODEL = 'cross-encoder/ms-marco-MiniLM-L-6-v2'
TOP_K = 10  # candidates for re-ranking
FINAL_TOP_K = 3  # final top-k after re-ranking

print("Loading embedding model...")
bi_encoder = SentenceTransformer(EMBEDDING_MODEL)

print("Loading cross-encoder for re-ranking...")
cross_encoder = CrossEncoder(CROSSENCODER_MODEL)

print("Downloading dataset from Hugging Face Hub...")
dataset = load_dataset("FreedomIntelligence/RAG-Instruct", split="train", trust_remote_code=True)

questions_dataset = [item["question"] for item in dataset]
print(f"✅ {len(questions_dataset)} questions loaded.")

# Generate embeddings once
print("🧠 Generating embeddings...")
embeddings_dataset = bi_encoder.encode(questions_dataset, convert_to_tensor=True)

def retrieve_and_rerank(user_question):
    # Encode user question
    user_embedding = bi_encoder.encode(user_question, convert_to_tensor=True)

    # Initial retrieval via cosine similarity
    similarities = util.pytorch_cos_sim(user_embedding, embeddings_dataset)[0]
    top_k = torch.topk(similarities, k=TOP_K)
    top_k_indices = top_k.indices.tolist()
    top_k_scores = top_k.values.tolist()

    # Build pairs for re-ranking with cross-encoder
    candidate_pairs = [(user_question, questions_dataset[i]) for i in top_k_indices]

    print("🔁 Re-ranking with CrossEncoder...")
    rerank_scores = cross_encoder.predict(candidate_pairs)

    # Combine original index with reranked score
    reranked = list(zip(top_k_indices, rerank_scores))
    reranked.sort(key=lambda x: x[1], reverse=True)  # sort by rerank score

    # Final Top-K after reranking
    final_results = []
    for idx, score in reranked[:FINAL_TOP_K]:
        q = dataset[idx]["question"]
        documents = dataset[idx]["documents"]
        if isinstance(documents, str):
            documents = [documents]
        final_results.append({
            "question": q,
            "rerank_score": round(float(score), 4),
            "documents": documents
        })

    return final_results

# %%
# 🧪 INTERACTIVE TEST
user_question = input("Enter your question: ")

results = retrieve_and_rerank(user_question)

print("\n🔎 Final Top-K Questions (after reranking):\n")
for i, item in enumerate(results, 1):
    print(f"🔹 Rank #{i}")
    print(f"🎯 Re-rank Score: {item['rerank_score']}")
    print(f"❓ Question: {item['question']}")
    print(f"📄 Documents:")
    for doc in item["documents"]:
        print(f"   • {doc.strip()[:300]}...")  # Preview
    print("-" * 80)
```
