```python
import numpy as np
import os
from sklearn.metrics.pairwise import cosine_similarity, linear_kernel
import csv

model_names = [
    "multi-qa-mpnet-base-dot-v1",
    "multi-qa-distilbert-dot-v1",
    "multi-qa-MiniLM-L6-dot-v1",
    "multi-qa-mpnet-base-cos-v1",
    "multi-qa-distilbert-cos-v1",
    "multi-qa-MiniLM-L6-cos-v1",
    "msmarco-bert-base-dot-v5",
    "msmarco-distilbert-dot-v5",
    "msmarco-distilbert-base-tas-b",
    "msmarco-distilbert-cos-v5",
    "msmarco-MiniLM-L12-cos-v5",
    "msmarco-MiniLM-L6-cos-v5"
]

base_path = "/kaggle/input"  
documents_per_question = 10
output_file = "embedding_score_file.csv"
```

```python
import os
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity, linear_kernel

output_npy_file = "embedding_scores.npy"
all_rankings = []

for model_name in model_names:
    print(f"\nProcessing model: {model_name}")

    q_path = os.path.join(base_path, model_name.lower(), f"st_qsts_embeddings_train-{model_name}.npy")
    d_path = os.path.join(base_path, model_name.lower(), f"st_docs_embeddings_train-{model_name}.npy")

    # Load embeddings
    q_emb = np.load(q_path)
    d_emb = np.load(d_path)

    num_questions = q_emb.shape[0]

    for i in range(num_questions):
        q_vec = q_emb[i].reshape(1, -1)
        
        start_idx = i * documents_per_question
        end_idx = (i + 1) * documents_per_question
        d_vecs = d_emb[start_idx:end_idx]
        
        cosine_scores = cosine_similarity(q_vec, d_vecs)[0]
        
        # Create list of (score, local_doc_index)
        scores_with_idx = [(cosine_scores[j], j) for j in range(documents_per_question)]
        
        # Sort by score descending
        sorted_scores = sorted(scores_with_idx, key=lambda x: x[0], reverse=True)

        all_rankings.append(sorted_scores)

# Save as .npy
np.save(output_npy_file, np.array(all_rankings, dtype=object))
print(f"\nFile '{output_npy_file}' correctly saved.")
```
